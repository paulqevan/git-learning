```python
def quicksort(li):
    if len(li) < 2 :
        return li
    else :
        pivot = li[0]
        petits = [k for k in li[1:] if k <= pivot]
        grands = [k for k in li[1:] if k > pivot]
        return quicksort(petits) + [pivot] + quicksort(grands)
    
```

```python
graph = {}
graph["vous"] = ["alice", "bob", "claire"]
graph["bob"] = ["anuj", "peggy"]
graph["alice"] = ["peggy"]
graph["claire"] = ["thom", "johnny"]
graph["anuj"] = []
graph["thom"] = []
graph["johnny"] = []
graph["peggy"] = []
```

```python
graph
```

```python
G = nx.DiGraph(graph)

# 3. Définition de la disposition (layout) des nœuds
# L'algorithme 'spring' écarte les nœuds pour limiter les chevauchements
pos = nx.spring_layout(G, seed=42) 

# 4. Personnalisation du tracé
plt.figure(figsize=(10, 8))  # Taille de la fenêtre (largeur, hauteur)

# Dessiner les nœuds
nx.draw_networkx_nodes(G, pos, 
                       node_size=2500,        # Taille des cercles
                       node_color="#66c2a5",  # Couleur Vert émeraude pastel
                       edgecolors="black")    # Contour des cercles

# Dessiner les arêtes (liens avec flèches)
nx.draw_networkx_edges(G, pos, 
                       width=2,               # Épaisseur des lignes
                       edge_color="#888888",  # Couleur Gris
                       arrowsize=20)          # Taille des flèches

# Dessiner les étiquettes (noms des personnes)
nx.draw_networkx_labels(G, pos, 
                        font_size=12,         # Taille du texte
                        font_family="sans-serif", 
                        font_weight="bold")   # Texte en gras

# 5. Affichage final
plt.title("Visualisation du réseau social de 'vous'", fontsize=16)
plt.axis("off")  # Masquer les axes X et Y pour un rendu propre
plt.show()
```

```python
from collections import deque
```

```python
def person_is_seller(name) :
    return name[-1] == 'm'
```

```python
def search(name):
    search_queue = deque()
    search_queue += graph[name]
    searched = []
    while search_queue :
        person = search_queue.popleft()
        if not person in searched: 
            if person_is_seller(person):
                print(person + " is a mango seller!")
                return True
            else :
                search_queue += graph[person]
                searched.append(person)
    return False
```

```python
search("vous")
```

```python
def groupAnna(li):
    hash = {}
    for k in li :
        key = "".join(sorted(k))
        hash.setdefault(key, []).append(k)
    return list(hash.values())
        
mots = ["eat", "tea", "tan", "ate", "nat", "bat"]
```

```python
def longest(nums):
    num_set =  set(nums)
    max_len = 0

    for k in num_set :
        if (k-1) not in num_set :
            current_num = k
            current_streak = 1

            while current_num + 1 in num_set:
                current_num +=1
                current_streak +=1

            max_len = max(max_len, current_streak)
    return(max_len)
    
```

```python
def topKFrequent(nums: List[int], k: int) -> List[int]:
    occurences = Counter(nums)
    top_k = occurences.most_common(k)
    return [num for num, _ in occurences.most_common(k)]
```
